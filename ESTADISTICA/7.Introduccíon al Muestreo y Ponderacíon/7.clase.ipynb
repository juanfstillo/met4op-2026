## Muestreo
### Necesidad del muestreo
Entre las decisiones metodológicas previas al relevamiento está la de observar todas las unidades del universo —un censo— o sólo una parte representativa —una muestra—. Salvo en poblaciones pequeñas, se impone la muestra: el tiempo, el costo y la complejidad de la recolección, el procesamiento y el análisis descartan en la práctica relevar el universo completo. De ahí que el muestreo sea una técnica de aplicación casi general en la investigación social, y que conocer sus principios sea condición del análisis riguroso.

### Conceptos básicos
* Unidad de Análisis: Es la unidad acerca de la cual se solicita información, son los sujetos de los cuales decimos algo, se refiere a "de
quien" se esta hablando.
* Universo o Población: Es el conjunto de unidades de análisis que constituyen un area de interes analítico, cuyo estudio buscamos. Lo que contituye la Población, depende de la definición del Problema de Investigación y del enfoque teórico del mismo.
* Observación de la Población: Existen dos formas de observar a una Población, a traves de un Censo o mediante una Muestra.
> CENSO: es el recuento total y simultaneo de todas las unidades de análisis que constituyen una Población. Es total porque una vez defindas todas las unidades deben ser consultadas y es simultaneo porque todas las consultas deben ser realizadas al mismo tiempo; esto no significa que deben realizarse en el mismo instante, sino durante un período de tiempo en el cual no se alteren las características de las unidades consultadas. Hacemos un Censo no solamente cuando analizamos la totalidad de una Población a nivel nacional o Provincial, sino cuando analizamos la totalidad de cualquier población, por ejemplo, Censo de escolaridad de un barrio determinado en una determinada ciudad o Censo de empleados de una determinada empresa. Los principales censos Nacionales son:el de Población, Familia y Vivienda, el Agropecuario y el Económiсо.

  > MUESTRA: es un subconjunto de unidades de análisis seleccionadas con la intención de averiguar varias características sobre la Población de la cual estan tomadas. Su objeto es obtener, por camino inferencial, conclusiones válidas para la población, partiendo de la observación de una parte pequeña de ésta.

* Marco muestral: Conjunto de listados, archivos, cartograf´ıa, etc. a partir de los cuales podemos seleccionar una muestra probabilística
Los marcos de muestreo no tiene porqué contener un listado de todas las unidades finales de muestreo
* Parámetro vs estadístico: Un parámetro es una medida numérica que describe una población entera, mientras que un estadístico describe una muestra. La diferencia principal radica en la fuente de los datos: el parámetro usa a todos los individuos de la población y el estadístico usa solo a un subconjunto.
* Esperanza vs sesgo:
> * La esperanza es el valor promedio esperado de un estimador
> * El sesgo es la diferencia entre esa esperanza y el valor real que se quiere medir
> * Un estimador es insesgado si su valor esperanza coincide con el parámetro a estimar


### Tipos de muestreos
* **Muestreo probabilísitico:**  

Es un proceso de selección de muestra en el cual las unidades son elegidas por métodos aleatorios. Existen variaciones en el muestreo probabilístico, pero todas comparten un rasgo común: *la selección de las unidades para la muestra se realiza por procedimiento al azar y con probabilidades conocidas de selección.*
**Una muestra es probabilística si todo elemento del universo U tiene una probabilidad positiva y conocida de ser seleccionado.**
 En base a ello se pueden hacer Inferencias (estimar los valores de la Población), calcular el margen de error y establecer con precisión el tamaño óptimo de la muestra. Cuando decimos una "probabilidad conocida" no quiere decir
"probabilidad igual". Tipos de muestras probabilísticas:

> * Muestreo Aleatorio Simple
> * Muestreo Sistemático
> * Muestreo Estratificado
> * Muestreo por Conglomerado



* **Muestreo no probabilístico:**  

Incluye todos los métodos en que las unidades no se seleccionan por procedimientos al azar o con probabilidades conocidas. La selección de unidades para que formen parte de la muestra se basa en gran parte en el criterio del investigador o el entrevistador de campo. Al no poder conocerse el error de muestreo no podemos hacer inferencias. Tipos de muestras no probabilísticas:

> * Muestreo Accidental
> * Muestreo Intencional
> * Muestreo por Cuotas
> * Muestreo de bola de nieve

### Muestreo probabilístico
El muestreo probabilístico tiene 2 ventajas importantes:
* Ausencia de sesgos de selección
* Posibilidad de estimar medidas de precisión: CV, intervalos de confianza, etc.

Estas dos ventajas desaparecen con la imposibilidad de medir lo que nos interesa a la unidad seleccionada (no respuesta, error de medición, etc)

#### Probabilidades de Selección
En la práctica no interesa la probabilidad $P(s)$ de seleccionar una muestra $s$ concreta, sino la probabilidad de que un **elemento** $i$ resulte seleccionado.

**Probabilidad de selección de primer orden** — que el elemento $i$ esté en la muestra:

$$ \pi_i = \sum\limits_{s \ni i} P(s) $$

**Probabilidad de selección de segundo orden** — que los elementos $i$ y $j$ estén ambos en la muestra:

$$ \pi_{ij} = \sum\limits_{s \ni i,j} P(s) $$

**Factor de expansión** — el inverso de la probabilidad de primer orden:

$$ F_i = \frac{1}{\pi_i} $$


#### Propiedades

Estas probabilidades cumplen dos propiedades que conviene tener presentes

**1) La suma de las probabilidades de inclusión coincide con el tamaño esperado de la muestra:**

$$ \sum_{i \in U} \pi_i = E(n(s)) $$

Esto se deduce de la propia definición de $\pi_i$: cada muestra $s$ que contiene a $i$ aporta $P(s)$ a $\pi_i$, de modo que al sumar sobre todos los $i \in U$, cada muestra $s$ termina contando $n(s)$ veces (una por cada uno de sus elementos):

$$
\begin{aligned}
\sum_{i \in U} \pi_i &= \sum_{i \in U} \sum_{s \ni i} P(s) \\
&= \sum_{s} n(s) \cdot P(s) \\
&= E(n(s))
\end{aligned}
$$

En un diseño de **tamaño fijo** (como la MAS que vemos a continuación), $n(s) = n$ para toda muestra, así que esta suma es simplemente $n$.

**2) El factor de expansión, sumado sobre la muestra efectivamente seleccionada, estima el tamaño de la población:**

$$ \sum_{i \in s} \frac{1}{\pi_i} \approx N $$

Cada unidad seleccionada "representa" a $F_i = 1/\pi_i$ unidades de la población; por eso, sumar los factores de expansión de los elementos que caen en la muestra permite reconstruir (estimar) el tamaño total $N$ del universo. Es la misma lógica que más adelante se usa para estimar totales y medias poblacionales a partir de la muestra.

> **Ejemplo.** Sea $U = \{\text{Lucía}, \text{Martín}, \text{Sofía}\}$ y consideremos el diseño que selecciona, con igual probabilidad, una de las siguientes tres muestras de tamaño 2:
>
> $$ S = \big\{ \{\text{Lucía}, \text{Martín}\},\ \{\text{Lucía}, \text{Sofía}\},\ \{\text{Martín}, \text{Sofía}\} \big\} $$
>
> $$ n(s) = 2, \qquad \text{card}(S) = 3 $$
>
> Al ser equiprobables, $P(s) = 1/3$ para cada una de las muestras $s$.
>
> Cada persona aparece en 2 de las 3 muestras posibles, así que su probabilidad de inclusión es:
>
> $$ \pi_{\text{Lucía}} = \pi_{\text{Martín}} = \pi_{\text{Sofía}} = \frac{2}{3} $$
>
> Verificamos la primera propiedad:
>
> $$ \sum_{i \in U} \pi_i = 3 \cdot \frac{2}{3} = 2 = E(n(s)) $$
>
> ya que $n(s) = 2$ en cada muestra. El factor de expansión de cada persona es:
>
> $$ F_i = \frac{1}{\pi_i} = \frac{3}{2} $$
>
> y, tomando por ejemplo la muestra $\{\text{Lucía}, \text{Martín}\}$, se verifica también la segunda propiedad: $\sum_{i \in s} F_i = \frac{3}{2} + \frac{3}{2} = 3 = N$.

> **Ejercicio.** De un curso con siete alumnos se desea seleccionar una $\text{MAS}(3)$ para una encuesta.
>
> $$ U = \{\text{Juan}, \text{Pedro}, \text{Maria}, \text{Tito}, \text{Lila}, \text{Carlos}, \text{Carla}\} $$
>
> Hallar:
> 1. $N$
> 2. $n$
> 3. $\text{card}(S)$ (cantidad de muestras posibles)
> 4. $P(s)$ (probabilidad de selección de cada muestra)
> 5. $\pi_i$ (probabilidad de selección de cada alumno)
> 6. $F_i$ (factor de expansión de cada alumno)

#### 1. Muestreo Aleatorio Simple
Estas muestras se definen como aquellas en las que cada unidad de la Población y toda combinación de unidades tienen la misma probabilidad de estar incluídas en la Muestra elegida.

¿Cómo seleccionar una MAS?
* Ordenar 'aleatoriamente' las N unidades
* Seleccionar las n primeras (o las n últimas)


##### Probabilidades de selección y factores de expansión en el MAS

$$ \pi_i = \frac{n}{N} $$

$$ F_i = \frac{N}{n} $$

#### Un estimador importante: el estimador de Horvitz-Thompson

$$ \widehat{t}_{\pi y} = \sum_{i \in s} \frac{Y_i}{\pi_i} \qquad \text{(estimador insesgado del total de } Y \text{)} $$

Análogamente, dividiendo por $N$ se obtiene un estimador insesgado de la media:

$$ \widehat{\bar{t}}_{\pi y} = \frac{1}{N} \cdot \sum_{i \in s} \frac{Y_i}{\pi_i} \qquad \text{(estimador insesgado de la media de } Y \text{)} $$

Nótese que esto es exactamente $\sum_{i \in s} F_i \cdot Y_i$: se pondera cada observación por su factor de expansión, la misma lógica ya vista para estimar $N$.

#### Propiedades deseables de un estimador

- **Insesgadez**: que el promedio (ponderado) de los valores posibles del estimador coincida con el parámetro a estimar.
- **Poca varianza**: que haya poca dispersión entre los distintos valores que toma el estimador.
- **Poco error medio cuadrático (EMC)**: que los distintos valores del estimador estén cerca del valor a estimar.

$$ \Rightarrow \text{Estimador acurado} $$

#### Esperanza, sesgo, varianza y error medio cuadrático de un estimador

Estas propiedades se formalizan promediando sobre todas las muestras $s$ posibles, ponderadas por $P(s)$:

**Esperanza y sesgo:**

$$ E(\widehat\theta) = \sum_{s} \widehat\theta(s) \cdot P(s) $$

$$ \text{Sesgo}(\widehat\theta) = E(\widehat\theta) - \theta $$

**Varianza y desvío estándar:**

$$ V(\widehat\theta) = \sum_{s} \big(\widehat\theta(s) - E(\widehat\theta)\big)^2 \cdot P(s) $$

$$ DS(\widehat\theta) = \sqrt{V(\widehat\theta)} $$

**Error medio cuadrático:**

$$ EMC(\widehat\theta) = \sum_{s} \big(\widehat\theta(s) - \theta\big)^2 \cdot P(s) = V(\widehat\theta) + \text{Sesgo}(\widehat\theta)^2 $$

Un estimador es **insesgado** si su esperanza coincide con el parámetro a estimar para cualquier vector $Y_i = Y_{i1}, \ldots, Y_{iN}$:

$$ E(\widehat\theta) = \theta $$

es decir, si su sesgo es cero. En ese caso, el EMC se reduce directamente a la varianza.

En el muestreo aleatorio simple:
* $\bar{y}$ es un estimador **insesgado** de la media poblacional $\bar{Y}$
* $N \cdot \bar{y}$ es un estimador insesgado del total poblacional $\sum\limits_{i=1}^{N} Y_i$
* $s^2$ es un estimador insesgado de $S^2$
* $s_{xy}$ es un estimador insesgado de $S_{xy}$

##### Caso importante: variables dicotómicas

En las encuestas una de las estimaciones más usuales son las de proporciones: qué proporción de unidades están dentro de cierta categoría $C$, o qué cantidad de unidades están en $C$.

Al estimar proporciones o totales aparecen variables que toman los valores $0$ y $1$: las **variables dicotómicas**.

$$
Y_i =
\begin{cases}
1, & \text{si } i \in C \\
0, & \text{si } i \notin C
\end{cases}
$$

Es usual escribir   
$P = \bar{Y}$,  
$Q = 1-P$,  
$p = \bar{y}$,  
y se cumple:  

$$ S^2 = P \cdot Q \cdot \frac{N}{N-1} \approx P \cdot Q $$

$$ \widehat{V}(p) = \left(1 - \frac{n}{N}\right) \cdot p \cdot q \cdot \frac{1}{n-1} $$

$$ CV(p) \approx \sqrt{\dfrac{Q}{n \cdot P}} $$

(omitiendo el factor de corrección por población finita)

El **coeficiente de variación** de un estimador es una medida de precisión que muestra el error estándar del estimador dividido entre el valor de la estimación, expresado usualmente como porcentaje. Sirve para saber qué tan confiable o estable es un cálculo estadístico.
$$ \text{CV} = 100 \cdot \frac{\text{DS}(\hat{\theta})}{\theta} \qquad (\theta \neq 0) $$

> **Ejemplo.** Si $\theta = 120$ y $\text{DS}(\hat{\theta}) = 12$:

#### 2. Muestreo Sistemático
El muestreo sistemático es un tipo de muestreo probabilístico donde se hace una selección aleatoria del primer elemento para la muestra, y luego se seleccionan los elementos posteriores utilizando intervalos fijos o sistemáticos hasta alcanzar el tamaño de la muestra deseado.

Un muestreo sistemático $\text{Sist}(I)$ queda definido por un único parámetro: el **intervalo de selección** $I$, un número entero con $0 < I \leq N$.

El procedimiento se reduce a dos pasos:

1. **Arranque aleatorio.** Se elige al azar, de forma equiprobable, un elemento de partida $a$ entre los primeros $I$ de la lista: $1 \leq a \leq I$.
2. **Salto fijo.** A partir de $a$, se selecciona **uno de cada $I$ elementos**: $a,\; a+I,\; a+2I,\; \dots$

Es decir: el azar interviene una sola vez —al elegir el arranque $a$—; a partir de ahí, toda la muestra queda determinada por el salto de tamaño $I$.

> Ejemplo: $N = 10$, $\text{Sist}(3)$
>
> Con una población de $N = 10$ elementos y un intervalo de selección $I = 3$, el arranque $a$ sólo puede tomar los valores $1$, $2$ o $3$. Cada arranque genera una muestra distinta, y **no hay más que tres posibles**:
>
> $$
\begin{aligned}
a = 1 \;\Rightarrow\; s_1 &= \{1, 4, 7, 10\} \\
a = 2 \;\Rightarrow\; s_2 &= \{2, 5, 8\} \\
a = 3 \;\Rightarrow\; s_3 &= \{3, 6, 9\}
\end{aligned}
$$
>
> Como el arranque se elige de forma equiprobable entre $1$, $2$ y $3$, **las tres muestras son equiprobables**: cada una tiene probabilidad $1/3$ de resultar seleccionada.

##### Aplicaciones del muestreo sistemático

En el trabajo de campo, el muestreo sistemático es la forma habitual de seleccionar unidades cuando existe un orden natural —físico o registral— que puede recorrerse a intervalos fijos:

- **Selección de viviendas dentro de un radio censal.**
- **Selección de viviendas en una manzana** (tomar una cada $I$ a lo largo del recorrido).
- **Selección de personas a partir de un registro** (padrón, listado de afiliados, base de clientes).
- **Control de calidad** (inspeccionar una pieza cada $I$ en una línea de producción).

#### 3. Muestreo Estratificado
La estratificación es el proceso de subdividir el Universo en subgrupos o estratos a fin de llevar a cabo procesos separados de Muestreo en cada uno de ellos.

Para este tipo de muestreo se divide a la Población en función de las categorías (estratos) de alguna variable que este correlacionada con la variable que queremos estudiar. La información censal nos permite conocer los tamaños de los estratos en la Población y poder estratificar en función de las variables incluídas en los listados (sexo, edad, area de residencia, ocupación, etc.). Los estratos pueden estar basados en un Criterio Único, como por ejemplo, la variable sexo, o en la Combinación de dos o mạs criterios (sexo y edad).

Hay dos criterios fundamentales para seleccionar las unidades que formaran la muestra dentro de cada estrato:
1) Estratificación Proporcional: Al número de unidades en cada estrato.
2) Estratificación no Proporcional: Esta es proporcional no al número de unidades de cada estrato sino a la variabilidad u homogeneidad en cada estrato. Esto se fundamenta en el hecho de que distribuciones que tengan un número pequeño de unidades son en general, mas heterogéneas que aquellas que tienen un gran número de elementos. O sea, cuanto mayor es la cantidad de unidades de una población, mas se homogeiniza, es decir, su desvió estándar es menor.
> Por esta razón, el estrato mas pequeño se verá representado en la muestra con un número algo superior a lo que le correspondería si se hiciera proporcional al número de unidades en la Población.

##### Algunos objetivos de la estratificación

Estratificar es dividir la población en grupos internamente homogéneos —los **estratos** (región, sexo, tamaño de localidad)— y muestrear dentro de cada uno. Se busca con ello:

- **Garantizar un tamaño mínimo en ciertos dominios**, para que ningún subgrupo de interés quede con demasiado pocos casos para estimar.
- **Reducir la varianza de los estimadores**, al eliminar la variabilidad *entre* estratos: el diseño estratificado suele ser más preciso que un aleatorio simple del mismo tamaño.
- **Asegurar una buena distribución de la muestra**, no necesariamente proporcional al tamaño de cada estrato.

##### Ejemplos de estratificaciones

Los estratos suelen definirse cruzando dos o más variables asociadas a lo que se estudia:

- **Muestra de empresas:** rama de actividad y tamaño.
- **Muestra de radios censales:** departamento y nivel educativo del jefe de hogar.
- **Muestra de alumnos:** carrera, sexo y tramo de edad.

##### Estratificación en la práctica: selección y estimación de varianzas

Cuando hay muchos estratos, tanto la **selección de la muestra** como la **estimación de varianzas** (CV, Deff, intervalos de confianza) se realizan con software estadístico —SPSS, R (paquete `survey`), etc.—, ya que las fórmulas se vuelven inmanejables a mano.

Es usual, además, **probar distintas alternativas de estratificación** antes de decidir la definitiva, evaluándolas sobre bases simuladas o de operativos anteriores para ver cuál reduce más la varianza.

Definida la estratificación del marco y seleccionada la muestra, se asigna a cada unidad su **factor de expansión** —la inversa de su probabilidad de selección, $F_i = 1/\pi_i$—, que es lo que permite luego realizar las estimaciones.

#### 3. Muestreo por conglomerados

En los diseños anteriores las unidades eran individuos. Aquí la lógica es distinta: la población se considera compuesta por **grupos**, cada uno con más de una unidad, y ese grupo recibe el nombre de **conglomerado** (*cluster*). El muestreo por conglomerados es aquel en el que la unidad de muestreo es el conglomerado, no la unidad individual.

Los conglomerados pueden ser **naturales** —cursos de un colegio, grupos de trabajo, manzanas, pueblos— o **artificiales**, formados por el investigador según necesidades prácticas.

##### Conglomerado monoetápico

Se selecciona una muestra de conglomerados y se encuesta a **todas** las unidades de los conglomerados seleccionados.

> *Ejemplo.* Seleccionamos una muestra de colegios (conglomerado de alumnos de sexto grado) y encuestamos, en los colegios elegidos, a **todos** los alumnos de sexto grado.

##### Conglomerado polietápico

Se selecciona una muestra de conglomerados y, dentro de cada uno, se encuesta sólo a una **muestra aleatoria** de sus unidades (no a todas).

> *Ejemplo.* Seleccionamos una muestra de radios censales (conglomerado de viviendas particulares) y, en cada radio elegido, seleccionamos a su vez una muestra aleatoria de viviendas. Encuestamos luego a todos los hogares de esas viviendas.

##### ¿Cuándo se selecciona una muestra por conglomerados?

- **Ausencia de un listado de las unidades finales de muestreo.** A menudo no existe un marco con todos los individuos (no hay una lista de todas las personas de una ciudad), pero sí de los conglomerados que los contienen (radios censales, manzanas). Se muestrean esos grupos y recién dentro de ellos se llega a las unidades finales.
- **Disminuir el costo del trabajo de campo.** Concentrar el relevamiento en unos pocos conglomerados —unas manzanas, unos colegios— reduce traslados y tiempos frente a una muestra dispersa por todo el territorio.

##### Ejemplos de conglomerados

- **Radios censales:** conglomerados de viviendas.
- **Escuelas:** conglomerados de alumnos.
- **Aulas:** conglomerados de alumnos.
- **Hogares:** conglomerados de personas.
- **Viviendas:** conglomerados de personas.

##### Diferencia entre estratificación y muestreo por conglomerados

Gráficamente, la diferencia (en general, aunque no siempre):

- **Estratificación ⇒ disminución de la varianza.** La población se divide en estratos internamente homogéneos, y de *cada uno* se toman casos. La muestra recorre todos los grupos.
- **Conglomeración ⇒ aumento de la varianza.** Se seleccionan *algunos* conglomerados enteros y quedan afuera los demás. La muestra se concentra en unos pocos grupos.





#### Muestreo polietápico

En el muestreo por conglomerados es usual que, en lugar de relevar todas las unidades de los conglomerados seleccionados, se **subseleccionen** unidades dentro de ellos. Cuando esa subselección se hace en dos o más niveles, el diseño se llama **polietápico**. Los conglomerados suelen estar **anidados**, y cada nivel de ese anidamiento define una etapa de selección:

- Radios censales ⇒ Viviendas
- Radios censales ⇒ Manzanas ⇒ Viviendas
- Colegio ⇒ Aulas
- Radios censales ⇒ Viviendas ⇒ Hogar ⇒ Persona
- Región (NEA, NOA, …) ⇒ Provincia ⇒ Radio censal

##### El caso bietápico

El polietápico más simple es el **bietápico**. El universo se divide en subuniversos llamados **Unidades de Primera Etapa (UPE)**; mediante algún diseño se selecciona una muestra de UPE; y dentro de cada UPE seleccionada se toma, a su vez, una muestra de sus elementos, en forma independiente de una UPE a otra.

##### El factor de expansión final

Como la selección de un elemento atraviesa varias etapas, su probabilidad final de ser elegido resulta de combinar la de cada etapa. En consecuencia, el **factor de expansión final** de cada elemento se obtiene combinando los factores de expansión de todas las etapas por las que pasó. Ese factor final es el que indica a cuántas unidades de la población representa cada caso relevado, y es precisamente el **peso de diseño** con el que arranca la ponderación: la probabilidad de selección que se fija al diseñar la muestra es la que después alimenta el peso.

### Muestreo no probabilístico

Al no poder conocerse el error de muestreo no podemos hacer inferencias probabilísticas

#### Muestreo Accidental o Causal
Este tipo de muestras se seleccionan de acuerdo con la conveniencia del investigador. Se toman los casos que "vienen a la mano", continuando el proceso hasta que la muestra adquiere el tamaño deseado.

#### Muestreo Intencional
Es una técnica no probabilística donde el investigador elige de forma deliberada a los participantes o casos según sus características específicas y los objetivos del estudio.

#### Muestreo por Cuotas
El objetivo básico del muestreo por cuota es la selección de una muestra que sea una réplica de la población que se quiere generalizar. La condición básica no es que los distintos estratos (sexo, edad, educación, nivel económico social, etc.) sean representados por muestras en sus proporciones correctas, sino mas bien que haya suficiente número de casos de cada estrato para hacer posible una estimación del valor del estrato en la población y que conozcamos la proporción en que cada estrato se constituye en la población total.

#### Muestreo de bola de nieve
Se utiliza cuando los participantes potenciales son difíciles de encontrar o si la muestra está limitada a un subgrupo muy pequeño de la población. En este tipo de muestreo los participantes de una investigación reclutan a otros participantes para una prueba o estudio.

El método de muestreo de bola de nieve se basa en referencias de sujetos iniciales para generar sujetos adicionales, de ahí deriva su nombre “bola de nieve”.

## Del muestreo a la necesidad de ponderar

### Tres conjuntos que no coinciden

- **Población objetivo:** el universo sobre el que queremos concluir (por ejemplo, el electorado de la Ciudad de Buenos Aires).
- **Marco muestral:** la lista o mecanismo del que efectivamente extraemos casos (un padrón, un panel online, un directorio telefónico).
- **Muestra observada:** los casos que finalmente responden.

Cada salto introduce error:

| Salto | Error asociado | Ejemplo |
|---|---|---|
| Población → marco | **Error de cobertura** | Un marco telefónico deja fuera a hogares sin línea. |
| Marco → muestra seleccionada | **Error de muestreo** | Azar de la selección; se cuantifica con el margen de error. |
| Muestra seleccionada → respondentes | **Error de no respuesta** | Los jóvenes o ciertos segmentos responden menos. |

### El rol de la ponderación

Si los segmentos que sobre-representamos difieren sistemáticamente en la variable de interés (por ejemplo, si los jóvenes responden más *y* votan distinto), la muestra estará **sesgada**. La ponderación reasigna la "voz" de cada respondente para restaurar el peso poblacional de cada segmento.

En términos estrictos, ponderar es asignar a cada caso de la muestra un peso $ w_i $ que expresa a cuántas unidades de la población representa ese respondente. La idea es sencilla: si un grupo quedó sobre-representado en la muestra, sus casos deben pesar menos; si quedó sub-representado, deben pesar más. Con los pesos adecuados, la muestra ponderada recupera la composición de la población en las variables que se usan para ajustar.

#### Qué corrige y qué no

Conviene fijar con precisión el alcance del instrumento. La ponderación reconstruye la composición poblacional en las variables que se emplean para ajustar, y sólo en ellas. La representatividad no se fabrica retrospectivamente; un buen diseño muestral sigue siendo insustituible.

Además, la reconstrucción tiene un costo. Al apartar los pesos de la unidad se introduce variabilidad adicional en las estimaciones: pesos muy dispares equivalen a apoyarse en menos casos efectivos. De ahí que toda ponderación deba evaluarse (a través del efecto de diseño y del tamaño de muestra efectivo) y que el analista enfrente siempre un compromiso entre sesgo y varianza: se acepta perder algo de precisión a cambio de una maqueta más fiel, cuidando de no pagar por la corrección más varianza de la necesaria.

## Estrategias de ponderación: de la post-estratificación al raking

### Post-estratificación

Si conocemos la **distribución conjunta** completa de las variables de ajuste en la población —por ejemplo, la tabla cruzada *Sexo × Grupo etario × Región* del Censo—, podemos formar celdas y asignar a cada una un peso que iguale su participación muestral a la poblacional. Es exacto y transparente.

**Su limitación práctica** es doble:

1. **Exige la distribución conjunta completa, que rara vez está disponible.** La post-estratificación forma celdas cruzando todas las variables de ajuste entre sí, y necesita conocer la participación poblacional de cada combinación: no basta con saber qué proporción de la población es varón y qué proporción tiene entre 30 y 49 años, sino que hace falta el dato de qué proporción es varón, de 30 a 49 años, del AMBA y con secundario completo, y así para cada casillero. El problema es que las fuentes oficiales —el Censo, la EPH— publican habitualmente los marginales (la distribución por sexo, por edad, por región, por separado) y a lo sumo algún cruce de a dos variables, pero casi nunca la tabla conjunta de todas las dimensiones que uno querría usar para ponderar. Sin esa tabla, sencillamente no se puede calcular el objetivo poblacional de cada celda, y el método queda inaplicable.
   * *Ejemplo:* Se desea ponderar por sexo (2 categorías), edad (4), región (4) y nivel educativo (4). La post-estratificación reclama la proporción poblacional de las $2 × 4 × 4 × 4 = 128$ combinaciones. El Censo ofrece cómodamente los cuatro marginales; la tabla de 128 celdas cruzadas, en cambio, no suele estar calculada.
2. **La proliferación de celdas genera casilleros vacíos o casi vacíos, y con ellos pesos inestables.** Aun suponiendo que se dispusiera de la tabla conjunta, la cantidad de celdas crece como el producto del número de categorías, no como su suma. En el ejemplo anterior son 128 celdas para una muestra de, por ejemplo, 1.500 casos: un promedio de menos de doce casos por celda, pero repartidos de manera muy despareja. Las combinaciones poco frecuentes caerán en la muestra con cero, uno o dos respondentes. Cuando una celda tiene un peso poblacional apreciable pero un solo caso muestral, todo ese peso descansa sobre un único individuo: su respuesta se magnifica de manera desproporcionada y la estimación se vuelve inestable (alta varianza). Y si la celda directamente no tiene casos, no hay a quién reponderar: el casillero queda vacío y el estimador, indefinido para ese segmento.

### El problema que resuelve el raking

El *raking* fue concebido precisamente para sortear ambos obstáculos: opera sobre los marginales —la información que las fuentes sí publican, lo que resuelve la primera limitación— y ajusta las distribuciones margen por margen en lugar de exigir que cada celda interior esté poblada, con lo que tolera los casilleros ralos o vacíos que descarrilan a la post-estratificación.

Con frecuencia conocemos únicamente los **totales marginales** de la población:

- distribución poblacional por **sexo** (Varón 48 % / Mujer 52 %),
- distribución poblacional por **grupo etario**,
- distribución poblacional por **región**,

pero **no** el cruce simultáneo de las tres. El **raking** permite construir pesos que reproducen *todos* esos marginales a la vez, sin exigir la distribución conjunta.

## El raking (*iterative proportional fitting*)

### La intuición

El raking parte de los márgenes que queremos alcanzar: la distribución poblacional de sexo, la de edad, la de región. Ajusta los pesos para cuadrar el primero; al hacerlo descoloca un poco los demás. Ajusta el segundo, y vuelve a mover levemente los ya cuadrados. Pero cada vuelta deja un desvío menor que la anterior, hasta que todos los márgenes coinciden a la vez con sus objetivos. Ese *ir y venir hasta la estabilidad* es el raking. De ahí su nombre técnico: **ajuste proporcional iterativo**.

### El algoritmo

El raking modifica los pesos de la muestra para que sus totales marginales coincidan con los totales de control de la población. El término *raking* sugiere una analogía con el proceso de alisar la tierra en un huerto o jardín, trabajando alternativamente de un lado a otro con un rastrillo en dos direcciones perpendiculares.

El procedimiento ajusta una variable de control por vez. Se toma la primera —por ejemplo, sexo— y se multiplican los pesos de cada categoría por el cociente entre su total poblacional y su total muestral ponderado, de modo que su marginal quede alineado con el de la población. Eso descoloca los marginales de las demás variables, así que se repite la operación sobre la siguiente —por ejemplo, edad—, y así sucesivamente según las variables por las que se pondere. Cada nueva pasada vuelve a mover levemente lo ya cuadrado, pero el desajuste se achica en cada vuelta, y tras unas iteraciones todos los marginales concuerdan a la vez. El resultado conserva la asociación entre las variables tal como aparece en la muestra. Con tres o más variables el procedimiento es idéntico: se recorre una a una cada dimensión de control y se repite el ciclo hasta converger.

Lo ideal es aplicar el raking sobre variables asociadas tanto a los resultados clave de la encuesta como a la falta de respuesta o de cobertura: es esa doble relación la que reduce el sesgo. En la práctica median otras consideraciones —una variable como el sexo, aunque tenga poca relación con ambas, puede incluirse igual para preservar la validez aparente de la muestra—.

### Categorías ralas: colapso y agrupación

El raking tolera celdas *conjuntas* vacías —esa es justamente su ventaja sobre la post-estratificación—, pero **no tolera una categoría vacía en un margen que se le pide igualar**. Si a una categoría de control no le corresponde ningún respondente en la muestra, su factor de ajuste queda indefinido: hay un objetivo poblacional que repartir, pero no hay sobre quién hacerlo (una división por cero). Y aun sin llegar al caso extremo, una categoría con muy pocos casos genera un factor de ajuste grande que recae sobre un puñado de respondentes: sus pesos se disparan, la varianza crece y el algoritmo puede oscilar sin converger.

La solución habitual es **colapsar**: agrupar las categorías demasiado finas en otras más amplias, de modo que cada una reúna una cantidad razonable de casos antes del raking. En lugar de ponderar por las 24 jurisdicciones del país —varias de las cuales aparecerían con dos o tres respondentes—, se agrupan en un puñado de **regiones** (AMBA, Centro, NOA-NEA, Cuyo, Patagonia); en lugar de edades año a año, en **tramos** (16-29, 30-49, 50-64, 65+). 

> **Ejemplo.** Se quiere ponderar por región usando las 24 jurisdicciones. Tierra del Fuego representa el 0,3 % de la población y en una muestra de 1.500 casos cae con 4 respondentes. El objetivo poblacional obliga a cargar sobre esos 4 casos todo el peso de la jurisdicción, con factores de ajuste elevados y una estimación regional sostenida por cuatro respuestas. Colapsando Tierra del Fuego dentro de "Patagonia", la categoría pasa a tener suficientes casos y el peso se reparte de manera estable.

El costo es el previsible: al agrupar, se iguala a la población en un margen **más grueso**, y se pierde la corrección fina que daría el detalle. Pero es un intercambio favorable —cuadrar de manera robusta un margen amplio vale más que cuadrar un margen detallado con pesos desbocados—.

### Diagnósticos de la ponderación: Deff, cv y ESS

Ponderar corrige sesgo, pero **tiene un costo en varianza**: cuanto más dispares son los pesos, menos información efectiva aporta la muestra. Antes de recortar conviene medir esa dispersión, y hay tres indicadores que van de la mano.

El **coeficiente de variación de los pesos** ($\text{cv}$) es la medida básica de su dispersión: el desvío estándar de los pesos dividido por su media. Con pesos idénticos $\text{cv}=0$; cuanto más desiguales, mayor es.

El **efecto de diseño por ponderación** (*design effect*, Deff) traduce esa dispersión en pérdida de eficiencia. En la formulación de Kish:

$$ \text{Deff} = \frac{n\sum_i w_i^2}{\left(\sum_i w_i\right)^2} = 1 + \text{cv}^2 $$

El Deff es simplemente $1$ más el cuadrado del coeficiente de variación de los pesos. Un $\text{Deff}=1{,}20$ significa que la ponderación infló la varianza de las estimaciones un 20 % respecto de una muestra aleatoria simple del mismo tamaño.

El **tamaño de muestra efectivo** (*effective sample size*, ESS) expresa ese costo en la unidad más intuitiva —cantidad de casos—:

$$ n_{\text{ef}} = \frac{n}{\text{Deff}} $$

Con $\text{Deff}=1{,}20$ y $n=1.500$, el ESS es $1.500/1{,}20 = 1.250$: la muestra ponderada rinde, en precisión, como una aleatoria simple de 1.250 casos. Se pagó la corrección del sesgo con el equivalente a 250 casos de precisión.

### Recorte de pesos (*trimming*)

El raking puede converger y, aun así, arrojar pesos con **fuerte variabilidad**. Esa dispersión se mide de dos formas habituales: la **razón entre el peso máximo y el mínimo**, y el **efecto de diseño por ponderación**, $1 + \text{cv}^2$, donde $\text{cv}$ es el coeficiente de variación de los pesos. Pesos muy dispares inflan la varianza de las estimaciones —sobre todo en subgrupos, donde unos pocos casos con peso desbocado dominan el resultado—.

El **recorte** acota esa dispersión en los dos extremos: los pesos por debajo de un mínimo $L$ se suben a $L$, y los que superan un máximo $U$ se bajan a $U$ (en encuestas, el problema dominante suele ser el techo). Ese tope se fija como un múltiplo del promedio —convencionalmente **3×** o **5x**—.

El criterio que justifica el recorte es la reducción del **error cuadrático medio** (MSE) de las estimaciones clave. Como $\text{MSE} = \text{varianza} + \text{sesgo}^2$, y el recorte **baja la varianza pero introduce algo de sesgo**, sólo conviene cuando la caída de varianza supera al aumento del sesgo al cuadrado. Es, por lo tanto, un compromiso deliberado y cuantificable.

El trimming es un tema distinto del raking, pero íntimamente ligado a él: suele ser el **último paso** del cálculo de pesos, que en general es el propio raking. De ahí una consecuencia técnica: al topear, la suma de los pesos deja de coincidir con el total poblacional, por lo que hay que **normalizar**.

### Normalización de los pesos

El recorte introduce una operación que reaparece de forma recurrente: la **normalización**. Los pesos que devuelve el raking tienen una escala arbitraria —importan sus proporciones relativas, no su magnitud absoluta—, y normalizar es reescalarlos para que **sumen $n$** (el tamaño de la muestra), con lo cual su **promedio queda en 1**:

$$ w_i^{\text{norm}} = w_i \cdot \frac{n}{\sum_j w_j} $$

Lo primero que conviene subrayar es que **esto no altera ninguna estimación**. Como el estimador ponderado es un cociente —$\sum_i w_i y_i \big/ \sum_i w_i$—, multiplicar todos los pesos por una misma constante se cancela entre numerador y denominador. La normalización es, entonces, puramente interpretativa, y cumple tres funciones: que cada peso se lea como **influencia relativa al caso medio** (1 = caso promedio; 1,20 = un 20 % por encima; 0,87 = un 13 % por debajo); que las fórmulas de diagnóstico —efecto de diseño, $n$ efectivo— queden expresadas de forma limpia; y que un umbral como el "3 veces la media" del recorte tenga un significado inmediato y comparable entre encuestas.

## Implementación con `balance`

Definimos los **marginales poblacionales objetivo** y aplicamos el raking sobre la muestra. Usamos la función `rake()` de la librería `balance`, que devuelve un **peso por fila**. *(El código asume un `df` de encuesta ya cargado, con las columnas de las variables para ponderar.)*

```python
import pandas as pd
import numpy as np
import logging
logging.getLogger("balance").setLevel(logging.ERROR)
import warnings
warnings.filterwarnings("ignore")
from balance.weighting_methods.rake import rake, prepare_marginal_dist_for_raking

df = pd.read_csv("data/ejemplo_ponderacion.csv")

# Recategorizar edad continua a tramos (necesarios para el raking)
df['edad_cat'] = pd.cut(
    df['edad'],
    bins=[16, 29, 44, 59, np.inf],
    labels=['16-29', '30-44', '45-59', '60+'],
    right=True,
    include_lowest=True,
)

# Marginales poblacionales objetivo (Censo 2022)
TARGETS = {
    'sexo': {'femenino': 0.53, 'masculino': 0.47},
    'edad_cat': {'16-29': 0.29, '30-44': 0.29, '45-59': 0.21, '60+': 0.21},
    'region': {'Región Metropolitana': 0.42, 'Región Pampeana': 0.23, 'Región NOA': 0.14,
               'Región NEA': 0.08, 'Región Cuyo': 0.07, 'Región Patagonia': 0.06},
    'nivel_educativo': {'prim': 0.29, 'sec': 0.44, 'terc': 0.12, 'univ': 0.13, 'pos': 0.02},
}
vars_rake = list(TARGETS)

print(f"{len(df)} casos cargados")
print(df['edad_cat'].value_counts(dropna=False).sort_index())
df.head()
```

```python
# Las variables de ajuste deben ser texto para evitar desajustes de tipo
for var in vars_rake:
    df[var] = df[var].astype(str)

df['peso'] = 1   # a priori, todos los casos pesan 1

# Distribucion marginal objetivo en el formato que espera balance
target_df = prepare_marginal_dist_for_raking(TARGETS)
target_weights = pd.Series(1.0, index=target_df.index, name="w_target")

# Raking: devuelve un peso por fila
res = rake(
    sample_df=df[vars_rake],
    sample_weights=df['peso'],
    target_df=target_df,
    target_weights=target_weights,
    variables=vars_rake,
)
df['peso'] = res['weight'].values
df.head()
```

```python
media_pesos = df['peso'].mean()
piso  = media_pesos / 3
techo = media_pesos * 3

print(f"Piso  (media / 3): {piso:.3f}")
print(f"Techo (media × 3): {techo:.3f}")
```

```python
def trimming_pesos(pesos, factor=3):
    promedio = np.mean(pesos)
    return np.clip(pesos, promedio / factor, promedio * factor)   # piso y techo

df['peso'] = trimming_pesos(df['peso'])

df.head()
```

```python
def normalizar_pesos(pesos):
    return pesos / pesos.sum() * len(pesos)   # reescala a que sumen n (media 1)

print(f"Suma total antes de normalizar: {df['peso'].sum():.1f}")
print(f"Peso promedio antes de normalizar: {df['peso'].mean():.3f}")

df['peso'] = normalizar_pesos(df['peso'])
df.head()
```

```python
print(f"Suma total: {df['peso'].sum():.1f}") #post normalización
print(f"Peso promedio: {df['peso'].mean():.3f}")
```

```python
sin_ponderar = df['voto'].value_counts(normalize=True) * 100
ponderada = df.groupby('voto')['peso'].sum() / df['peso'].sum() * 100
tabla_voto = pd.DataFrame({'sin ponderar %': sin_ponderar.round(1), 'ponderada %': ponderada.round(1)}) \
               .sort_values('ponderada %', ascending=False)
print(tabla_voto)
```
